```python
from datasets import load_dataset

# Cargar los datos de las imagenes del Dataset
dataset = load_dataset("mohanty/PlantVillage", "default")

print(dataset)  #comprobación de que se ha cargado correctamente

```

##### Primeros 10 datos, modificando su visualización, quitando espacios y eliminando dupliciades

```python
import pandas as pd
columnas = []

for split in dataset:
    primeros = dataset[split].select(range(min(10, len(dataset[split]))))

    for posicion, registro in enumerate(primeros):
        columnas.append({
            "Conjunto(train/test)": split,
            "Fotografía": str(registro["text"]).strip()
        })



tabla_hojas = pd.DataFrame(columnas, columns=["Conjunto(train/test)", "Fotografía"])
tabla_hojas["Fotografía"] = tabla_hojas["Fotografía"].str.strip().str.title()  #elimina espacios y poner la primera en mayúscula
tabla_hojas = tabla_hojas.drop_duplicates(subset="Fotografía", keep="first").reset_index(drop=True)  #elimina duplicidades  se queda con la primera opción

display(tabla_hojas)
```
